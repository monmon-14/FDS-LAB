```python
#Analyzing the trend of Job postings in Data Science
import pandas as pd
import matplotlib.pyplot as plt
data={'Year': list(range(2013,2025)),
      'Job Postings':[100,150,300,350,400,550,700,1100,2100,2500,3100,4100]}
df=pd.DataFrame(data)
plt.plot(df['Year'],df['Job Postings'],marker='o')
plt.title('Trend of Data Science Job Postings')
plt.xlabel('Year')
plt.ylabel('Number of Job Postings')
plt.show()
```

```python
#Analyzing the distribution of various data science
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data={
    'job_title':['Data Scientist','Data Analyst','Data Engineer',
    'Machine Learning', 'Data Scientist','Data Analyst']
}
df=pd.DataFrame(data)
sns.countplot(x='job_title',data=df)
plt.title('Distribution of Data Science Jobs')
plt.xlabel('Roles')
plt.ylabel('Count')
plt.show()
```

```python
#Analyzing the distribution of various data science
import pandas as pd
import matplotlib.pyplot as plt

roles = ['Data Analyst', 'Data Engineer', 'Data Scientist', 'ML Engineer',
'Business Analyst']
counts = [200,500,150,300,450]
plt.bar(roles, counts)
plt.title('Distribution of Data Science Roles')
plt.xlabel('Role')
plt.ylabel('Count')
plt.show()
```
