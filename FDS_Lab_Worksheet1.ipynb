```python
#Q1
import pandas as pd
data={
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
    }
df=pd.DataFrame(data)
print(df)
```

```python
#Q2
df1=df[df['Age']>30]
print(df1)
```

```python
#Q3
df['Salary']=[50000,60000,70000,80000]
print(df)
```

```python
#Q4
df['Salary']=df['Salary']*1.10
print(df)
```

```python
#Q5
df=df.drop("City",axis=1)
print(df)
```

```python
df=df.rename(columns={'Age':'Years'})
print(df)
```

```python
#Q7
df2=df.sort_values(by='Salary',ascending=False)
print(df)
```

```python
#Q8
data = {
'Department': ['HR', 'IT', 'HR', 'IT'],
'Employee': ['Alice', 'Bob', 'Charlie', 'David'],
'Salary': [50000, 60000, 55000, 65000]
}
dept_df=pd.DataFrame(data)
avg_sal=dept_df.groupby("Department")['Salary'].mean()
print(avg_sal)
```

```python
#Q9
df1 = pd.DataFrame({
'ID': [1, 2, 3],
'Name': ['Alice', 'Bob', 'Charlie']
})
df2 = pd.DataFrame({
'ID': [1, 2, 4],
'Score': [85, 90, 88]
})
merge_df=pd.merge(df1,df2,on='ID')
print(merge_df)
```
