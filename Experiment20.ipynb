```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# New dataset
data = {
    "Annual_Income": [15, 16, 18, 20, 22, 35, 38, 40, 42, 45,
                      60, 62, 65, 68, 70, 80, 85, 88, 90, 95],
    "Spending_Score": [80, 85, 75, 90, 82, 60, 65, 55, 62, 58,
                       40, 45, 35, 42, 38, 20, 25, 18, 22, 15]
}

df = pd.DataFrame(data)

print("Dataset:")
print(df)

print("\nDataset Information:")
df.info()

print("\nFirst 5 Rows:")
print(df.head())

# Select features
features = df[["Annual_Income", "Spending_Score"]].values

# Create K-Means model
model = KMeans(n_clusters=5, random_state=42, n_init=10)

# Train the model
model.fit(features)

# Add cluster labels
df["Cluster"] = model.predict(features)

print("\nClustered Dataset:")
print(df)

# Plot clusters
plt.figure(figsize=(8, 6))

plt.scatter(
    df["Annual_Income"],
    df["Spending_Score"],
    c=df["Cluster"],
    s=100
)

plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("K-Means Clustering")
plt.grid(False)
plt.show()
```
