```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = {
    "total_bill": [16.99, 10.34, 21.01, 23.68, 24.59, 18.50, 15.20, 30.50, 22.10, 12.50],
    "tip": [1.01, 1.66, 3.50, 3.31, 3.61, 2.50, 2.00, 4.50, 3.00, 1.50],
    "sex": ["Female", "Male", "Male", "Male", "Female", "Male", "Female", "Male", "Female", "Male"],
    "smoker": ["No", "No", "No", "No", "No", "Yes", "No", "Yes", "No", "Yes"],
    "day": ["Sun", "Sun", "Sun", "Sun", "Sun", "Sat", "Sat", "Sat", "Fri", "Fri"],
    "time": ["Dinner", "Dinner", "Dinner", "Dinner", "Dinner", "Dinner", "Dinner", "Dinner", "Lunch", "Lunch"],
    "size": [2, 3, 3, 2, 4, 3, 2, 4, 2, 2]
}

tips = pd.DataFrame(data)

print("Dataset:")
print(tips)

# Quantitative Analysis
print("\nQuantitative Analysis:")
print(tips.describe())

# Qualitative Analysis
print("\nQualitative Analysis:")
print(tips["time"].value_counts())
print(tips["day"].value_counts())

# Distribution of Total Bill
plt.hist(tips["total_bill"], bins=5, edgecolor="black")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.title("Distribution of Total Bill")
plt.show()

# Tip vs Total Bill
plt.scatter(tips["tip"], tips["total_bill"])
plt.xlabel("Tip")
plt.ylabel("Total Bill")
plt.title("Tip vs Total Bill")
plt.show()

# Tip vs Total Bill with trend line
plt.scatter(tips["tip"], tips["total_bill"])

m, b = np.polyfit(tips["tip"], tips["total_bill"], 1)
plt.plot(tips["tip"], m * tips["tip"] + b)

plt.xlabel("Tip")
plt.ylabel("Total Bill")
plt.title("Tip vs Total Bill with Regression Line")
plt.show()

# Box plot
plt.boxplot(tips["total_bill"])
plt.ylabel("Total Bill")
plt.title("Box Plot of Total Bill")
plt.show()

# Bar chart for Time
tips["time"].value_counts().plot(kind="bar")
plt.xlabel("Time")
plt.ylabel("Count")
plt.title("Customers by Time")
plt.show()
```
