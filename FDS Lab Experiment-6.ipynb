```python
# Experiment to detect outliers in a given data set
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Given data
array = np.array([27, 50, 44, 6, 58, 61, 23, 86,
                  67, 20, 75, 7, 79, 61, 90, 54])
print("Given Array:")
print(array)

# Mean
print("\nMean:")
print(array.mean())

# Percentiles
print("\n25th Percentile (Q1):")
print(np.percentile(array, 25))
print("\n50th Percentile (Q2 / Median):")
print(np.percentile(array, 50))
print("\n75th Percentile (Q3):")
print(np.percentile(array, 75))
print("\n100th Percentile:")
print(np.percentile(array, 100))

# Function for outlier detection
def outDetection(array):
    Q1, Q3 = np.percentile(array, [25, 75])
    IQR = Q3 - Q1
    lr = Q1 - (1.5 * IQR)
    ur = Q3 + (1.5 * IQR)
    return lr, ur

# Calculate lower and upper range
lr, ur = outDetection(array)
print("\nLower Range (LR):")
print(lr)
print("\nUpper Range (UR):")
print(ur)

# Plot original data
sns.histplot(array, kde=True)
plt.title("Original Data")
plt.show()

# Remove outliers
new_array = array[(array > lr) & (array < ur)]

print("\nArray after removing outliers:")
print(new_array)

# Plot after removing outliers
sns.histplot(new_array, kde=True)
plt.title("Data After Removing Outliers")
plt.show()

# Detect outliers again
lr1, ur1 = outDetection(new_array)
print("\nNew Lower Range:")
print(lr1)
print("\nNew Upper Range:")
print(ur1)

# Final array
final_array = new_array[(new_array > lr1) & (new_array < ur1)]
print("\nFinal Array:")
print(final_array)

# Final plot
sns.histplot(final_array, kde=True)
plt.title("Final Data")
plt.show()
```
